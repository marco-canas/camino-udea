## 1. Enunciado Corregido



> **Pregunta 62.** Ana y Juan hicieron $400$ pasabocas para una fiesta. Ana hacía $3$ pasabocas por minuto y Juan $2$ por minuto. Juan trabajó $25$ minutos más que Ana. Los tiempos, en minutos, que trabajaron Juan y Ana respectivamente fueron:
> * **A.** $95$ y $70$
> * **B.** $90$ y $65$
> * **C.** $85$ y $60$
> * **D.** $80$ y $55$ 
> 



## 2. Estrategias de Solución Rápida (Menos de 2 minutos)



### Estrategia 1: Descartar por enunciado (La más rápida)

El enunciado pide explícitamente los tiempos de **Juan y Ana respectivamente**.
También dice: *"Juan trabajó 25 minutos más que Ana"*. Por lo tanto:




$$\text{Tiempo de Juan} > \text{Tiempo de Ana}$$



### Estrategia 2: Resolver quitando la ventaja inicial

1. **Calcular lo que hizo Juan a solas:** Como Juan trabajó $25$ minutos más a una tasa de $2$ pasabocas/minuto, él hizo solo:

$$25 \text{ min} \times 2 \text{ pasabocas/min} = 50 \text{ pasabocas}$$




2. **Restar esa producción al total:** Los pasabocas restantes se hicieron trabajando en simultáneo:

$$400 - 50 = 350 \text{ pasabocas}$$




3. **Tiempo compartido:** Trabajando juntos, hacen $3 + 2 = 5$ pasabocas por minuto:

$$v = \frac{pasabocas}{tiempo} \quad \Rightarrow \quad  \text{Tiempo de juan y Ana } = \frac{350}{5} = 70 \text{ minutos}$$




4. **Tiempo de Juan:** Sumamos sus $25$ minutos extra:

$$\text{Tiempo de Juan} = 70 + 25 = 95 \text{ minutos}$$



Al buscar la combinación $95$ y $70$ en la opción **A**, confirmamos que la relación matemática coincide perfectamente con la opción **A**.





## 3. Código en Python (Modelo, Tabla y Gráfico)

Este script contiene la función de simulación matemática, genera la tabla comparativa con `pandas` y grafica el avance de la producción con `matplotlib`.



````python
import matplotlib.pyplot as plt
import pandas as pd
````

````python


def simular_produccion_pasabocas(tiempo_ana):
    """Calcula la producción total de pasabocas basada en el tiempo de Ana."""
    tasa_ana = 3  # pasabocas/min
    tasa_juan = 2  # pasabocas/min
    ventaja_juan = 25  # minutos

    tiempo_juan = tiempo_ana + ventaja_juan

    # Producciones individuales
    prod_ana = tiempo_ana * tasa_ana
    prod_juan = tiempo_juan * tasa_juan
    prod_total = prod_ana + prod_juan

    # Creación de la estructura tabular
    datos = {
        "Persona": ["Ana", "Juan", "Total Combinado"],
        "Tasa (u/min)": [tasa_ana, tasa_juan, tasa_ana + tasa_juan],
        "Tiempo Trabajado (min)": [tiempo_ana, tiempo_juan, None],
        "Pasabocas Producidos": [prod_ana, prod_juan, prod_total],
    }

    df = pd.DataFrame(datos)
    return df, tiempo_juan
````

````python


# --- 1. EXPERIMENTACIÓN CON EL MODELO ---
# Probamos con el valor deducido: Ana = 70 minutos
tiempo_ana_optimo = 70
df_resultado, tiempo_juan_optimo = simular_produccion_pasabocas(
    tiempo_ana_optimo
)

print("=== REPRESENTACIÓN TABULAR (PANDAS) ===")
print(df_resultado.to_string(index=False, na_rep="-"))
````

````python


# --- 2. REPRESENTACIÓN GRÁFICA (MATPLOTLIB) ---
# Vamos a graficar cómo se acumulan los pasabocas minuto a minuto
minutos = list(range(0, tiempo_juan_optimo + 1))
pasabocas_juan = []
pasabocas_ana = []
pasabocas_totales = []

# Nota: Juan empieza a trabajar 25 minutos antes, o Ana termina 25 minutos antes.
# Asumiendo que terminan al mismo tiempo y Juan empezó 25 minutos antes:
for m in minutos:
    # Producción de Juan
    p_juan = m * 2
    pasabocas_juan.append(p_juan)

    # Producción de Ana (comienza en el minuto 25)
    p_ana = (m - 25) * 3 if m >= 25 else 0
    pasabocas_ana.append(p_ana)

    # Total acumulado
    pasabocas_totales.append(p_juan + p_ana)
````

````python
import numpy as np 
plt.figure(figsize=(10, 6))
plt.plot(minutos, pasabocas_juan, label="Producción Juan (2 u/min)", color="#1f77b4")
plt.plot(
    minutos, pasabocas_ana, label="Producción Ana (3 u/min desde min 25)", color="#ff7f0e"
)
plt.plot(
    minutos,
    pasabocas_totales,
    label="Producción Total Acumulada",
    color="#2ca02c",
    linestyle="--",
    linewidth=2,
)

# Línea de meta de los 400 pasabocas
plt.axhline(y=400, color="r", linestyle=":", label="Meta (400 pasabocas)")
plt.axvline(x=95, color="gray", linestyle=":")
plt.xticks(np.arange(0, 100, 5))
# Ajustes de la gráfica
plt.title("Modelo del Progreso de Producción de Pasabocas")
plt.xlabel("Tiempo total transcurrido (minutos)")
plt.ylabel("Cantidad de Pasabocas")
plt.legend(loc="upper left")
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.savefig(r"C:\Users\marco\Downloads\produccion_pasabocas.png")
plt.show()
````
