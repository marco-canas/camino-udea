### **Enunciado**

> **63.** Un número de dos cifras es tal que si estas se suman el resultado es 13 y si al número se le resta 27 las cifras se invierten. Entonces el producto de las cifras del número es:
> A. 30
> B. 32
> C. 36
> D. 40

---



### **Solución Matemática (en LaTeX)**

Sea el número de dos cifras representado como $N = 10x + y$, donde:

* $x$ es la cifra de las decenas ($1 \le x \le 9$)
* $y$ es la cifra de las unidades ($0 \le y \le 9$)



Del enunciado obtenemos las siguientes dos condiciones:

1. **La suma de las cifras es 13:**

$$x + y = 13 \quad \text{--- (Ecuación 1)}$$


2. **Al restar 27, las cifras se invierten ($10y + x$):**

$$(10x + y) - 27 = 10y + x$$




Simplificando esta ecuación:

$$10x - x + y - 10y = 27$$


$$9x - 9y = 27$$


Dividiendo entre 9 a ambos lados:

$$x - y = 3 \quad \text{--- (Ecuación 2)}$$



---

#### **Resolución del Sistema de Ecuaciones**

Sumamos la **Ecuación 1** y la **Ecuación 2**:


$$(x + y) + (x - y) = 13 + 3$$

$$2x = 16 \implies x = 8$$

Sustituyendo $x = 8$ en la **Ecuación 1**:


$$8 + y = 13 \implies y = 5$$

Por lo tanto, el número original es:


$$N = 85$$

#### **Producto de las cifras:**

$$\text{Producto} = x \cdot y = 8 \cdot 5 = 40$$

**Respuesta correcta:** **D. 40**

---



### **Solución con Python (NumPy, Pandas y Matplotlib)**



````python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# 1. Generación del espacio de búsqueda con NumPy y Pandas
numeros = np.arange(10, 100)
df = pd.DataFrame({"numero": numeros})

# Extraemos decenas y unidades
df["decenas"] = df["numero"] // 10
df["unidades"] = df["numero"] % 10

# Aplicamos las condiciones del problema
df["suma_cifras"] = df["decenas"] + df["unidades"]
df["numero_invertido"] = df["unidades"] * 10 + df["decenas"]
df["numero_resta_27"] = df["numero"] - 27

# Condición: suma de cifras == 13 Y número - 27 == número invertido
df["cumple"] = (df["suma_cifras"] == 13) & (
    df["numero_resta_27"] == df["numero_invertido"]
)

# Filtramos la solución
solucion = df[df["cumple"]].iloc[0]

x = int(solucion["decenas"])
y = int(solucion["unidades"])
producto = x * y

print(f"Número encontrado: {solucion['numero']}")
print(f"Cifra de las decenas (x): {x}")
print(f"Cifra de las unidades (y): {y}")
print(f"Producto de las cifras (x * y): {producto}")

# 2. Gráfica de las ecuaciones con Matplotlib
x_vals = np.arange(1, 10)
y_suma = 13 - x_vals  # De x + y = 13
y_resta = x_vals - 3  # De x - y = 3
````

````python

plt.figure(figsize=(7, 5))
plt.plot(
    x_vals,
    y_suma,
    label=r"Ecuación 1: $x + y = 13$",
    color="#1f77b4",
    linewidth=2,
    marker="o",
)
plt.plot(
    x_vals,
    y_resta,
    label=r"Ecuación 2: $x - y = 3$",
    color="#ff7f0e",
    linewidth=2,
    marker="s",
)

# Punto de intersección (solución)
plt.plot(x, y, "ro", markersize=10, label=f"Solución: (x={x}, y={y})")
plt.annotate(
    f"  N = {x}{y}\n  ({x} × {y} = {producto})",
    xy=(x, y),
    xytext=(x - 1.8, y + 0.5),
    arrowprops=dict(facecolor="black", shrink=0.05, width=1, headwidth=6),
    fontsize=10,
    fontweight="bold",
    bbox=dict(boxstyle="round,pad=0.3", fc="yellow", ec="b", lw=1),
)

plt.title(
    "Representación Gráfica del Sistema de Ecuaciones",
    fontsize=12,
    fontweight="bold",
)
plt.xlabel("Cifra de las decenas ($x$)", fontsize=11)
plt.ylabel("Cifra de las unidades ($y$)", fontsize=11)
plt.xlim(0, 10)
plt.ylim(0, 10)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(fontsize=10)
plt.tight_layout()
plt.savefig(r"C:\Users\marco\Downloads\sistema_ecuaciones.png")
plt.show()
````
