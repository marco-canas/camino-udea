
## 1. Enunciado Corregido

> **Pregunta 77.** Cinco amigos Andrés, Bernardo, Carlos, Darío y Ernesto juegan a pasarse un balón y lo hacen de la siguiente manera: Andrés se lo pasa a Darío, Bernardo se lo pasa a Andrés, Carlos se lo pasa a Bernardo, Darío se lo pasa a Ernesto y Ernesto se lo pasa a Carlos. Si al inicio del juego Darío tiene el balón, entonces al ser pasado $523$ veces, quien tiene finalmente el balón es:
> * **A.** Andrés
> * **B.** Bernardo
> * **C.** Carlos
> * **D.** Darío
> 




# 2. Estrategia de Solución Rápida (Menos de 2 minutos)

## Paso 1: Mapear el ciclo de pases

Sigamos la trayectoria del balón partiendo desde **Darío** (Pase 0):

* **Inicio (Pase 0):** Darío tiene el balón.
* **Pase 1:** Darío se lo pasa a **Ernesto**.
* **Pase 2:** Ernesto se lo pasa a **Carlos**.
* **Pase 3:** Carlos se lo pasa a **Bernardo**.
* **Pase 4:** Bernardo se lo pasa a **Andrés**.
* **Pase 5:** Andrés se lo pasa a **Darío**.



Al llegar al pase 5, el balón regresa a Darío. Esto significa que el juego forma un **ciclo perfecto de 5 pases**.

### Paso 2: Aplicar aritmética modular (El truco del residuo)

Dado que la secuencia se repite idénticamente cada 5 pases, dividimos el total de pases ($523$) entre el tamaño del ciclo ($5$) para encontrar el residuo:

$$523 \div 5 = 104 \text{ con un residuo de } 3$$

El residuo nos dice cuántos pasos adicionales se dan después de completar los ciclos enteros:

* Residuo $0 \rightarrow$ Darío
* Residuo $1 \rightarrow$ Ernesto
* Residuo $2 \rightarrow$ Carlos
* **Residuo $3 \rightarrow$ Bernardo**
* Residuo $4 \rightarrow$ Andrés

**Respuesta Correcta:** **B. Bernardo**



## 3. Código en Python (Modelo, Tabla y Gráfico)

El siguiente script define el modelo cíclico, genera un DataFrame descriptivo con `pandas` y visualiza la trayectoria del balón usando `matplotlib`.


````python
import matplotlib.pyplot as plt
import pandas as pd


def simular_pases_balon(pases_totales):
    """Calcula quién tiene el balón paso a paso en un ciclo cerrado."""
    # Definición del orden del ciclo partiendo de Darío (Pase 0)
    ciclo = ["Darío", "Ernesto", "Carlos", "Bernardo", "Andrés"]
    periodo = len(ciclo)

    # Listas para almacenar el histórico
    historial_pases = []
    historial_jugadores = []

    # Registrar el desarrollo paso a paso hasta el límite solicitado
    for i in range(pases_totales + 1):
        historial_pases.append(i)
        jugador_actual = ciclo[i % periodo]
        historial_jugadores.append(jugador_actual)

    # Crear Estructura Tabular con Pandas
    df_evolucion = pd.DataFrame(
        {"Número de Pase": historial_pases, "Poseedor del Balón": historial_jugadores}
    )

    return df_evolucion, ciclo


# --- 1. EXPERIMENTACIÓN CON EL MODELO ---
total_pases_pregunta = 523
df_resultado, orden_ciclo = simular_pases_balon(total_pases_pregunta)

print("=== MUESTRA DE LA REPRESENTACIÓN TABULAR (PANDAS) ===")
# Mostramos los primeros pases y los últimos pases para verificar
print(df_resultado.head(6).to_string(index=False))
print("...")
print(df_resultado.tail(4).to_string(index=False))

# Quien tiene el balón al final:
ganador = df_resultado.iloc[total_pases_pregunta]["Poseedor del Balón"]
print(f"\nAl pase {total_pases_pregunta}, el balón lo tiene: {ganador}")


# --- 2. REPRESENTACIÓN GRÁFICA (MATPLOTLIB) ---
# Graficaremos los primeros 15 pases para observar visualmente la naturaleza cíclica u onda periódica
df_grafico = df_resultado.head(16).copy()

# Mapeo numérico de los jugadores para el eje Y de la gráfica
mapeo_numerico = {jugador: idx for idx, jugador in enumerate(orden_ciclo)}
df_grafico["Y_Id"] = df_grafico["Poseedor del Balón"].map(mapeo_numerico)
````

````python

plt.figure(figsize=(10, 5))
plt.plot(
    df_grafico["Número de Pase"],
    df_grafico["Y_Id"],
    marker="o",
    linestyle="-",
    color="#1f77b4",
    linewidth=2,
    markersize=8,
)

# Configuración de los ejes y etiquetas
plt.yticks(range(len(orden_ciclo)), orden_ciclo)
plt.xticks(range(16))
plt.xlabel("Número de Pase")
plt.ylabel("Jugador con el Balón")
plt.title(
    "Naturaleza Cíclica del Juego de Pases (Primeros 15 pases)\nResiduo de 523 % 5 = 3 -> Bernardo"
)
plt.grid(True, linestyle="--", alpha=0.5)

# Resaltar un punto cualquiera con residuo 3 (ej. Pase 3 o Pase 13) como demostración visual
plt.annotate(
    "¡Mismo estado que pase 523! (Residuo 3)",
    xy=(3, mapeo_numerico["Bernardo"]),
    xytext=(5, mapeo_numerico["Bernardo"] - 0.5),
    arrowprops=dict(facecolor="black", shrink=0.05, width=1, headwidth=6),
)

plt.tight_layout()
plt.show()
````
