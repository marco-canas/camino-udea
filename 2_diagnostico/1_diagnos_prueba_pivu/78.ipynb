## 1. Enunciado Corregido

> **Pregunta 78.** Una persona que pesa $120\text{ kg}$ decide hacer una dieta. Esta persona sabe que el $40\%$ de su masa total está conformada por grasa. Si durante el primer mes de su dieta esta persona pierde el $60\%$ de su grasa y mantiene los otros índices iguales, entonces al finalizar el primer mes ella pesa:
> * **A.** $89{,}6\text{ kg}$
> * **B.** $91{,}2\text{ kg}$
> * **C.** $96\text{ kg}$
> * **D.** $100\text{ kg}$
> 
> 

---

## 2. Estrategias de Solución Rápida (Menos de 2 minutos)

### Estrategia 1: Cálculo directo por componentes (La clásica)

1. **Calcular la grasa inicial:** El $40\%$ de $120\text{ kg}$ es:

$$120 \times 0{,}40 = 48\text{ kg de grasa}$$


2. **Calcular la masa que NO es grasa (fija):** 
$$120 - 48 = 72\text{ kg}$$


3. **Calcular la grasa remanente:** Si pierde el $60\%$, le queda el $40\%$ de esa grasa:

$$48 \times 0{,}40 = 19{,}2\text{ kg de grasa}$$


4. **Peso final:** Sumamos la masa fija y la nueva grasa:

$$72 + 19{,}2 = 91{,}2\text{ kg}$$



### Estrategia 2: El "truco" del porcentaje de pérdida total (La más veloz)

1. Sabes que la grasa representa el $40\%$ del cuerpo y que se pierde el $60\%$ de ese porcentaje.
2. Multiplicas los porcentajes para saber cuánto peso total se perdió:

$$40\% \times 60\% = 24\% \text{ del peso total perdido}$$


3. Si pierde el $24\%$, se queda con el $76\%$ de su peso inicial ($100\% - 24\% = 76\%$).
4. Calculas el $76\%$ de $120\text{ kg}$:

$$120 \times 0{,}76 = 91{,}2\text{ kg}$$



**Respuesta Correcta:** **B. $91{,}2\text{ kg}$**

## 3. Código en Python (Modelo Matemático, Tabla y Gráfico)

Puedes usar el siguiente script en tu entorno de Python (asegúrate de tener `pandas` y `matplotlib` instalados).

```python


````python
import matplotlib.pyplot as plt
import pandas as pd


def simular_dieta(
    peso_inicial, pct_grasa_inicial, pct_grasa_perdida
):  # Modificado el nombre del argumento
    """Calcula la evolución del peso basado en la pérdida de grasa."""
    # Masa de grasa y masa magra inicial
    grasa_inicial = peso_inicial * (pct_grasa_inicial / 100.0)
    masa_magra = peso_inicial - grasa_inicial

    # Pérdida de grasa y grasa final
    grasa_perdida_kg = grasa_inicial * (pct_grasa_perdida / 100.0)
    grasa_final = grasa_inicial - grasa_perdida_kg

    # Peso final
    peso_final = masa_magra + grasa_final

    # Crear Diccionario de datos para estructura tabular
    datos = {
        "Componente": ["Masa Magra (No grasa)", "Grasa Corporal", "Total"],
        "Inicial (kg)": [masa_magra, grasa_inicial, peso_inicial],
        "Final (kg)": [masa_magra, grasa_final, peso_final],
        "Cambio (kg)": [0.0, -grasa_perdida_kg, -grasa_perdida_kg],
    }

    df = pd.DataFrame(datos)
    return df


# --- 1. EXPERIMENTACIÓN CON EL MODELO ---
peso_i = 120
porcentaje_grasa = 40
porcentaje_perdida = 60

df_resultado = simular_dieta(peso_i, porcentaje_grasa, porcentaje_perdida)

print("=== REPRESENTACIÓN TABULAR (PANDAS) ===")
print(df_resultado.to_string(index=False))
print(f"\nResultado del peso final: {df_resultado.iloc[2]['Final (kg)']} kg")

# --- 2. REPRESENTACIÓN GRÁFICA (MATPLOTLIB) ---
# Filtramos el total para graficar solo la composición del cuerpo
df_grafico = df_resultado[df_resultado["Componente"] != "Total"]

fig, ax = plt.subplots(figsize=(8, 5))

# Definición de barras apiladas
ax.bar(
    ["Estado Inicial", "Estado Final"],
    [df_grafico.iloc[0]["Inicial (kg)"], df_grafico.iloc[0]["Final (kg)"]],
    label="Masa Magra",
    color="#2ca02c",
    width=0.4,
)
ax.bar(
    ["Estado Inicial", "Estado Final"],
    [df_grafico.iloc[1]["Inicial (kg)"], df_grafico.iloc[1]["Final (kg)"]],
    bottom=[
        df_grafico.iloc[0]["Inicial (kg)"],
        df_grafico.iloc[0]["Final (kg)"],
    ],
    label="Grasa",
    color="#ff7f0e",
    width=0.4,
)

# Etiquetas de texto sobre las barras para ver los valores exactos
peso_inicial_total = df_resultado.iloc[2]["Inicial (kg)"]
peso_final_total = df_resultado.iloc[2]["Final (kg)"]
ax.text(
    0,
    peso_inicial_total + 2,
    f"{peso_inicial_total} kg",
    ha="center",
    weight="bold",
)
ax.text(
    1, peso_final_total + 2, f"{peso_final_total} kg", ha="center", weight="bold"
)

# Configuración estética del gráfico
ax.set_ylabel("Masa (kg)")
ax.set_title("Evolución de la Composición Corporal en el Primer Mes")
ax.set_ylim(0, 140)
ax.legend(loc="upper right")
ax.grid(axis="y", linestyle="--", alpha=0.7)

plt.tight_layout()
plt.show()
````
