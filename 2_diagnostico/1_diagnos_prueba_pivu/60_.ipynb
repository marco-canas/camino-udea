# Ejercicio 60 de la prueba diagnóstica del pivu  

**Enunciado de la pregunta**

> **60.** Se tienen dos cartas marcadas en cada una de sus caras con un número. La figura muestra uno de los números en cada una de las cartas (7 en la primera y 10 en la segunda). Los resultados de las sumas en todas las posibles posiciones en que pueden aparecer las cartas son 11, 12, 16 y 17. El número de parejas de números que se pueden colocar en las caras no visibles, que cumplen estas condiciones son:
> A. 1
> B. 2
> C. 4
> D. 9

---



**Solución Matemática (en LaTeX)**

Sean las cartas representadas como:

* Carta 1: $\{7, x\}$
* Carta 2: $\{10, y\}$

Donde $x$ e $y$ son los números en las caras no visibles.



Las cuatro sumas posibles combinando una cara de la primera carta con una cara de la segunda carta son:

1. $S_1 = 7 + 10 = 17$
2. $S_2 = 7 + y$
3. $S_3 = x + 10$
4. $S_4 = x + y$



El conjunto de sumas posibles dado en el enunciado es $\{11, 12, 16, 17\}$. Como la suma $7 + 10 = 17$ ya está en el conjunto, las otras tres sumas correspondientes a $\{7+y, x+10, x+y\}$ deben ser exactamente $\{11, 12, 16\}$.

Existen dos casos para asignar los valores de las sumas restantes $\{7+y, x+10\}$ entre los valores $\{11, 12, 16\}$:

* **Caso 1:**

$$7 + y = 11 \implies y = 4$$


$$x + 10 = 12 \implies x = 2$$





Verificamos la cuarta suma: $x + y = 2 + 4 = 6 \neq 16$. *(Inválido)*
Si probamos otra combinación en el Caso 1:

$$7 + y = 11 \implies y = 4$$


$$x + 10 = 16 \implies x = 6$$



Verificamos la cuarta suma: $x + y = 6 + 4 = 10 \neq 12$. *(Inválido)*
* **Caso 2:**

$$7 + y = 12 \implies y = 5$$


$$x + 10 = 11 \implies x = 1$$



Verificamos la cuarta suma: $x + y = 1 + 5 = 6 \neq 16$. *(Inválido)*
Probando la combinación alternativa:

$$7 + y = 12 \implies y = 5$$


$$x + 10 = 16 \implies x = 6$$



Verificamos la cuarta suma: $x + y = 6 + 5 = 11$.


Las cuatro sumas son $\{17, 12, 16, 11\}$, lo cual coincide perfectamente con $\{11, 12, 16, 17\}$.
Por lo tanto, la pareja de caras ocultas $(x, y) = (6, 5)$ es **válida**.
* **Caso 3:**

$$7 + y = 16 \implies y = 9$$


$$x + 10 = 11 \implies x = 1$$



Verificamos la cuarta suma: $x + y = 1 + 9 = 10 \neq 12$. *(Inválido)*
Probando la combinación alternativa:

$$7 + y = 16 \implies y = 9$$


$$x + 10 = 12 \implies x = 2$$



Verificamos la cuarta suma: $x + y = 2 + 9 = 11$.
Las cuatro sumas son $\{17, 16, 12, 11\}$, lo cual coincide con $\{11, 12, 16, 17\}$.
Por lo tanto, la pareja de caras ocultas $(x, y) = (2, 9)$ es **válida**.

Concluimos que existen **2 parejas** de números verdaderas para las caras no visibles: $(x,y) = (6, 5)$ y $(x,y) = (2, 9)$.

**Respuesta correcta: B. 2**

---



**Resolución con Python (NumPy, Pandas y Matplotlib)**


````python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# 1. Búsqueda y análisis de soluciones con NumPy y Pandas
target_sums = {11, 12, 16, 17}
valid_pairs = []

# Iteramos en un rango de posibles valores para las caras ocultas (x, y)
for x in range(-20, 30):
    for y in range(-20, 30):
        # Calculamos las 4 sumas posibles con NumPy
        sums = np.array([7 + 10, 7 + y, x + 10, x + y])
        if set(sums) == target_sums:
            valid_pairs.append({"x (Carta 1)": x, "y (Carta 2)": y, "Sumas": sorted(sums)})

# Convertimos a DataFrame de Pandas
df_solutions = pd.DataFrame(valid_pairs)
print("Parejas válidas encontradas:")
print(df_solutions)

# 2. Visualización con Matplotlib
fig, ax = plt.subplots(figsize=(6, 4))

# Graficamos todas las combinaciones y destacamos las soluciones
x_vals = np.arange(-5, 15)
y_vals = np.arange(-5, 15)
X, Y = np.meshgrid(x_vals, y_vals)

ax.scatter(X, Y, color="lightgray", alpha=0.5, label="Combinaciones evaluadas")

# Resaltamos los pares válidos
sol_x = df_solutions["x (Carta 1)"].values
sol_y = df_solutions["y (Carta 2)"].values

ax.scatter(sol_x, sol_y, color="red", s=100, zorder=5, label=f"Soluciones ({len(sol_x)})")

for i in range(len(sol_x)):
    ax.annotate(
        f"({sol_x[i]}, {sol_y[i]})",
        (sol_x[i], sol_y[i]),
        xytext=(sol_x[i] + 0.5, sol_y[i] + 0.5),
        fontweight="bold",
    )

ax.set_xlabel("Número oculto en Carta 1 (x)")
ax.set_ylabel("Número oculto en Carta 2 (y)")
ax.set_title("Soluciones válidas para las caras ocultas")
ax.grid(True, linestyle="--", alpha=0.6)
ax.legend()

plt.show()
````
