# Ejercicio 42 de la prueba final del PIVU

## Preguntas 41 y 42.

$A$, $B$ y $C$ son candidatos a la presidencia de una cooperativa.  

Una investigación mostró que, entre los socios entrevistados, 110 no van a votar.    

Entre los entrevistados que sí pretenden participar, 20 van a votar por el candidato A, 50 por el candidato B, 35 por el candidato $C$, 110 definitivamente no van a votar por el candidato $A$, 70 no van a votar por $C$, y 10 aun no han decidido entre $A$ y $C$.   

También, la investigación mostró que 15 de los entrevistados que pretenden votar, no tienen preferencia, y por tanto, pueden votar por cualquiera de los tres candidatos.   

Con base en estos datos:
41.	El número de entrevistados que no ha decidido si votar por $B$ o por $C$ es:

A.	0  
B.	15  
C.	25  
D.	30  
42.	El número de socios que respondió la entrevista es:

A.	110  
B.	270  
C.	265  
D.	160  

## Estrategia de solución

Uso de diagramas de ven para razonar la situación a través de conjuntos. 


## Solución:

````python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Datos del problema
a, b, c = 20, 50, 35
x, y, z = 0, 25, 10
i = 15
no_votan = 110

# Total de personas que pretenden votar
votan = a + b + c + x + y + z + i

# Total de entrevistados
total_entrevistados = votan + no_votan

print("Total votan:", votan)
print("Total no votan:", no_votan)
print("Total general:", total_entrevistados)

````

````python

Para determinar el número total de socios que respondieron la entrevista, se debe sumar el total de entrevistados que **pretenden votar** más los que **no van a votar**.

---

### 1. Desglose Matemático

A partir de los valores calculados en la pregunta anterior para cada uno de los subconjuntos del diagrama de Venn:

* **Solo por A ($a$):** 20
* **Solo por B ($b$):** 50
* **Solo por C ($c$):** 35
* **Indecisos solo entre A y B ($x$):** 0
* **Indecisos solo entre B y C ($y$):** 25
* **Indecisos solo entre A y C ($z$):** 10
* **Indecisos entre los tres A, B y C ($i$):** 15
* **Socios que NO van a votar:** 110

#### Paso 1: Sumar los socios que pretenden votar ($V$)

$$V = a + b + c + x + y + z + i$$

$$V = 20 + 50 + 35 + 0 + 25 + 10 + 15 = 155$$

#### Paso 2: Sumar al total de socios que no van a votar ($NV$)

$$\text{Total de Entrevistados} = V + NV = 155 + 110 = 265$$

---
````


### 2. Código en Python (NumPy, Pandas y Matplotlib)



````python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# ---------------------------------------------------------
# 1. Cálculo del Total con NumPy y Pandas
# ---------------------------------------------------------

# Valores de los subconjuntos
regiones_voto = {
    'Solo A': 20,
    'Solo B': 50,
    'Solo C': 35,
    'Indecisos A y B': 0,
    'Indecisos B y C': 25,
    'Indecisos A y C': 10,
    'Indecisos A, B y C': 15,
}

# Creación del DataFrame de votantes con Pandas
df_votan = pd.DataFrame(
    list(regiones_voto.items()), columns=['Grupo', 'Cantidad']
)

# Suma con NumPy
total_pretenden_votar = np.sum(df_votan['Cantidad'].to_numpy())
socios_no_votan = 110
total_entrevistados = total_pretenden_votar + socios_no_votan

# Estructura final en Pandas
df_resumen = pd.DataFrame(
    {
        'Estado': [
            'Pretenden Votar',
            'No van a Votar',
            'Total Socios Entrevistados',
        ],
        'Cantidad': [
            total_pretenden_votar,
            socios_no_votan,
            total_entrevistados,
        ],
    }
)

print("--- CONTEO GENERAL DE ENTREVISTADOS ---")
print(df_resumen.to_string(index=False))

# ---------------------------------------------------------
# 2. Visualización Global con Matplotlib
# ---------------------------------------------------------

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Gráfico 1: Desglose de los que pretenden votar
ax1.bar(df_votan['Grupo'], df_votan['Cantidad'], color='#2ecc71')
ax1.set_xticklabels(df_votan['Grupo'], rotation=45, ha='right')
ax1.set_title('Desglose de Socios que Pretenden Votar (155)')
ax1.set_ylabel('Cantidad')

for bar in ax1.patches:
    height = bar.get_height()
    ax1.annotate(
        f'{int(height)}',
        xy=(bar.get_x() + bar.get_width() / 2, height),
        xytext=(0, 3),
        textcoords='offset points',
        ha='center',
        va='bottom',
    )

# Gráfico 2: Composición Total de Entrevistados
ax2.pie(
    [total_pretenden_votar, socios_no_votan],
    labels=['Pretenden Votar (155)', 'No van a Votar (110)'],
    autopct='%1.1f%%',
    colors=['#3498db', '#e74c3c'],
    startangle=140,
    explode=(0.05, 0),
)
ax2.set_title(f'Composición Total: {total_entrevistados} Entrevistados')

plt.tight_layout()
plt.show()
````
