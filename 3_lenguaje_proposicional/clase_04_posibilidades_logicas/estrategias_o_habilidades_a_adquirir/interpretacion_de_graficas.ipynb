¡Claro! Esta es una excelente manera de practicar el razonamiento lógico-matemático. La clave aquí es entender que en un gráfico de **distancia vs. tiempo** respecto a un punto de referencia (como un museo o una casa):

* **Hacia el origen (eje x):** El objeto se acerca al punto de referencia.
* **Alejándose del origen:** El objeto se aleja.
* **Línea horizontal:** El objeto está quieto (la distancia no cambia).
* **Pendiente más inclinada:** Mayor velocidad.

Aquí tienes 5 conjuntos de problemas diseñados para entrenar esa habilidad:

---

## Conjunto 1: El trayecto al colegio
**Enunciado:** Los siguientes gráficos muestran la distancia de dos estudiantes, **A** y **B**, respecto a su colegio en función del tiempo. Ambos inician su recorrido al mismo tiempo.

1.  **Pregunta 1.1:** ¿Cuál gráfica ilustra que "A y B salieron de diferentes lugares, pero A caminó más rápido y alcanzó a B antes de llegar al colegio"?
2.  **Pregunta 1.2:** ¿Cuál gráfica muestra que "A vive más lejos que B, pero ambos llegaron al colegio exactamente al mismo tiempo"?
3.  **Pregunta 1.3:** ¿Cuál gráfica representa que "A se quedó esperando en la puerta del colegio mientras veía cómo B se alejaba de la institución"?

---

## Conjunto 2: Carreras en la pista
**Enunciado:** Dos corredores, **X** y **Y**, compiten en una pista recta. Los gráficos representan su distancia respecto a la **línea de meta** (donde distancia = 0 es la meta).

1.  **Pregunta 2.1:** "X comenzó con ventaja (más cerca de la meta), pero Y corrió tan rápido que llegó a la meta mucho antes que X".
2.  **Pregunta 2.2:** "Ambos comenzaron a la misma distancia de la meta, pero X se detuvo a mitad de camino por una lesión mientras Y continuó hasta el final".
3.  **Pregunta 2.3:** "X y Y empezaron en el mismo punto, mantuvieron la misma velocidad, pero Y decidió devolverse al punto de partida a mitad de la carrera".

---

## Conjunto 3: El nivel del tanque de agua
**Enunciado:** Se observa el nivel de agua (altura) en dos tanques, **C** y **D**, respecto al tiempo mientras se llenan o se vacían.

1.  **Pregunta 3.1:** "¿Qué gráfica muestra que el tanque C estaba vacío y empezó a llenarse, mientras el tanque D estaba lleno y empezó a vaciarse, cruzándose en el mismo nivel de agua en un punto dado?"
2.  **Pregunta 3.2:** "¿En qué gráfica se observa que el tanque C se llena el doble de rápido que el tanque D, partiendo ambos desde cero?"
3.  **Pregunta 3.3:** "El tanque C mantiene su nivel de agua constante porque la llave está cerrada, mientras que el tanque D pierde agua constantemente".

---

## Conjunto 4: Dos drones en el aire
**Enunciado:** Los gráficos representan la **altura** de dos drones, **P1** y **P2**, respecto al suelo.

1.  **Pregunta 4.1:** "P1 despega desde el suelo con velocidad constante, mientras que P2 desciende desde una altura considerable para aterrizar en el mismo instante que P1 alcanza su altura máxima".
2.  **Pregunta 4.2:** "Ambos drones se mantienen a una altura fija, pero P1 vuela a una altura mayor que P2 durante todo el trayecto".
3.  **Pregunta 4.3:** "P1 desciende hacia el suelo, pero a mitad de camino se detiene y queda suspendido en el aire (vuelo estacionario)".

---

## Conjunto 5: Los repartidores de comida
**Enunciado:** Un restaurante es el punto de referencia (distancia 0). Los gráficos muestran la posición de dos repartidores, **M** y **N**.

1.  **Pregunta 5.1:** "M sale del restaurante a entregar un pedido, mientras N regresa al restaurante después de haber entregado el suyo".
2.  **Pregunta 5.2:** "M y N salen del restaurante al mismo tiempo, pero N viaja en una moto más rápida y se aleja de M rápidamente".
3.  **Pregunta 5.3:** "M sale a entregar un pedido, pero olvida las llaves y debe regresar al restaurante inmediatamente, mientras N apenas está llegando al restaurante".

---

### Tips para resolver estos ejercicios:
* **Punto de intersección:** Significa que ambos están en el mismo lugar al mismo tiempo.
* **Corte con el eje horizontal (x):** Significa que el objeto llegó a su destino (distancia 0).
* **Corte con el eje vertical (y):** Indica la posición inicial (dónde estaban en el tiempo 0).

Para generar estos gráficos, utilizaremos subplots de **Matplotlib**. He diseñado el código para que cada bloque de 4 gráficos (similar a la imagen que subiste) corresponda a cada uno de los 5 conjuntos de preguntas.

Cada conjunto generará una figura con 4 opciones numeradas del (1) al (4), para que puedas realizar el ejercicio de emparejamiento.



```python
import matplotlib.pyplot as plt
import numpy as np
import re

def crear_graficos_ejercicios(titulo_figura, configuraciones):
    # Crear la figura
    fig, axs = plt.subplots(1, 4, figsize=(16, 5))
    fig.suptitle(titulo_figura, fontsize=16, fontweight='bold')
    
    t = np.linspace(0, 10, 100)
    
    for i, ax in enumerate(axs):
        # Configuración estética de los ejes
        ax.set_xlim(0, 10)
        ax.set_ylim(0, 10)
        ax.set_xlabel('Tiempo')
        ax.set_ylabel('Posición')
        ax.set_title(f'Opción ({i+1})')
        ax.set_xticks([])
        ax.set_yticks([])
        
        # Obtener funciones y etiquetas
        config = configuraciones[i]
        
        # Dibujar líneas
        ax.plot(t, config['p_func'](t), label=config['p_label'], color='blue', linewidth=2)
        ax.plot(t, config['q_func'](t), label=config['q_label'], color='red', linewidth=2)
        
        ax.legend(loc='upper right')
        ax.grid(True, linestyle='--', alpha=0.6)

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    
    # Limpiar el nombre del archivo (quitar ":" y espacios extra para evitar errores)
    nombre_archivo = re.sub(r'[^\w\s-]', '', titulo_figura).replace(' ', '_') + ".png"
    
    # Guardar la figura como imagen de alta calidad
    plt.savefig(nombre_archivo, dpi=300, bbox_inches='tight')
    print(f"Imagen guardada como: {nombre_archivo}")
    
    # Mostrar en pantalla y cerrar para limpiar memoria
    plt.show()
    plt.close(fig)

# --- CONFIGURACIONES DE LOS 5 CONJUNTOS ---

conjuntos = [
    ("Conjunto 1 - El trayecto al colegio", [
        {'p_label': 'A', 'q_label': 'B', 'p_func': lambda t: 10 - 1.5*t, 'q_func': lambda t: 7 - 0.5*t},
        {'p_label': 'A', 'q_label': 'B', 'p_func': lambda t: 10 - t, 'q_func': lambda t: 5 - 0.5*t},
        {'p_label': 'A', 'q_label': 'B', 'p_func': lambda t: 0*t, 'q_func': lambda t: 0.8*t},
        {'p_label': 'A', 'q_label': 'B', 'p_func': lambda t: 8 - t, 'q_func': lambda t: 8 - t}
    ]),
    ("Conjunto 2 - Carreras en la pista", [
        {'p_label': 'X', 'q_label': 'Y', 'p_func': lambda t: 5 - 0.4*t, 'q_func': lambda t: 10 - 2*t},
        {'p_label': 'X', 'q_label': 'Y', 'p_func': lambda t: np.where(t<5, 10-t, 5), 'q_func': lambda t: 10-t},
        {'p_label': 'X', 'q_label': 'Y', 'p_func': lambda t: 10-t, 'q_func': lambda t: np.where(t<5, 10-t, t)},
        {'p_label': 'X', 'q_label': 'Y', 'p_func': lambda t: 10-t, 'q_func': lambda t: 8-t}
    ]),
    ("Conjunto 3 - El nivel del tanque", [
        {'p_label': 'C', 'q_label': 'D', 'p_func': lambda t: t, 'q_func': lambda t: 10-t},
        {'p_label': 'C', 'q_label': 'D', 'p_func': lambda t: 2*t, 'q_func': lambda t: t},
        {'p_label': 'C', 'q_label': 'D', 'p_func': lambda t: 5 + 0*t, 'q_func': lambda t: 10-t},
        {'p_label': 'C', 'q_label': 'D', 'p_func': lambda t: t, 'q_func': lambda t: 0.5*t}
    ]),
    ("Conjunto 4 - Dos drones en el aire", [
        {'p_label': 'P1', 'q_label': 'P2', 'p_func': lambda t: t, 'q_func': lambda t: 10-t},
        {'p_label': 'P1', 'q_label': 'P2', 'p_func': lambda t: 8 + 0*t, 'q_func': lambda t: 4 + 0*t},
        {'p_label': 'P1', 'q_label': 'P2', 'p_func': lambda t: np.where(t<5, 10-t, 5), 'q_func': lambda t: 0*t},
        {'p_label': 'P1', 'q_label': 'P2', 'p_func': lambda t: 10-t, 'q_func': lambda t: 10-2*t}
    ]),
    ("Conjunto 5 - Los repartidores", [
        {'p_label': 'M', 'q_label': 'N', 'p_func': lambda t: t, 'q_func': lambda t: 10-t},
        {'p_label': 'M', 'q_label': 'N', 'p_func': lambda t: 0.5*t, 'q_func': lambda t: 2*t},
        {'p_label': 'M', 'q_label': 'N', 'p_func': lambda t: np.where(t<3, t, 6-t), 'q_func': lambda t: 10-t},
        {'p_label': 'M', 'q_label': 'N', 'p_func': lambda t: t, 'q_func': lambda t: 5 + 0*t}
    ])
]

# Ejecutar la creación para cada conjunto
for titulo, config in conjuntos:
    crear_graficos_ejercicios(titulo, config)
```

# Notas sobre el código:
* **Funciones Lambda:** Se usaron para definir los comportamientos rápidamente. Por ejemplo, `10 - t` representa algo que se acerca al origen, mientras que `t` representa algo que se aleja.
* **NumPy Where:** Lo usé en los casos donde hay cambios de dirección o paradas (como el repartidor que se devuelve o el dron que queda suspendido), para crear funciones lineales a trozos.
* **Estética:** Se eliminaron los números de los ejes (`xticks([])`) para que el estudiante se concentre en la **forma** de la pendiente y la relación visual, tal como en el examen de la Universidad de Antioquia que subiste.
