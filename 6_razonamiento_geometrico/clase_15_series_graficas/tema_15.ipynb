# Solución matemática del problema

Vamos a evaluar matemáticamente cada afirmación para determinar su veracidad:

* **Análisis de la Afirmación I:** *"A ingeniería ingresaron 800 aspirantes."*
* Geométricamente, el sector **I** representa un cuadrante completo del círculo de admitidos menos 22.5°, equivalente a un ángulo de $90^\circ - 22.5^\circ = 67.5$. 


* La proporción de admitidos de Ingeniería es: $\frac{67.5^\circ}{360^\circ} = 0.1875$ ($25\%$).


````python
67.5/360
````

* Número de admitidos en Ingeniería: $3600 \times 0.1875 = 675$.


````python
3600*0.1875
````

* Como $675 \neq 800$, la afirmación **I es FALSA**.




* **Análisis de la Afirmación II:** *"De cada 10 aspirantes a estudiar artes fueron admitidos 3."*
* El sector **A** también representa un cuadrante de $90^\circ$ ($25\%$).
* Número de admitidos en Artes: $3600 \times 0.25 = 900$.
* Del gráfico de barras sabemos que los aspirantes a Artes son 3 mil ($3000$).
* Relación de admisión para Artes: $\frac{\text{Admitidos}}{\text{Aspirantes}} = \frac{900}{3000} = \frac{3}{10}$.
* Esto significa exactamente que ingresan 3 de cada 10 aspirantes. Por lo tanto, la afirmación **II es VERDADERA**.


* **Análisis de la Afirmación III:** *"El porcentaje de admitidos con relación a aspirantes por programa es igual en artes que en derecho."*
* Ya calculamos la relación para Artes, la cual es del $30\%$.
* Para Derecho (**D**), el gráfico circular muestra otro cuadrante completo de $90^\circ$ ($25\%$), lo que equivale a $900$ admitidos.
* Del gráfico de barras vemos que Derecho tiene 4 mil ($4000$) aspirantes.
* Relación de admisión para Derecho: $\frac{900}{4000} = \frac{9}{40} = 0.225$ ($22.5\%$).
* Como $30\% \neq 22.5\%$, la afirmación **III es FALSA**.



Concluimos que **Solo la II es verdadera**. La respuesta correcta es la **(a)**.

## Representación del proceso de solución con Pandas y Plotly

Este script en Python utiliza **Pandas** para procesar las tablas de datos y **Plotly** para generar una visualización interactiva que compara directamente las tasas de admisión de los programas evaluados.



````python
import pandas as pd
import plotly.graph_objects as go

# 1. Crear el set de datos estructurado en un DataFrame de Pandas
datos = {
    'Programa': ['Ingeniería (I)', 'Artes (A)', 'Medicina (M)', 'Ciencias (C)', 'Derecho (D)'],
    'Aspirantes': [5000, 3000, 6000, 3000, 4000],
    'Ángulo Circular': [90-22.5, 90, 22.5, 22.5, 180 - 22.5] 
}

df = pd.DataFrame(datos)

# Calcular admitidos y tasa de admisión (porcentaje de ingreso por programa)
total_admitidos = 3600
df['Admitidos'] = (df['Ángulo Circular'] / 360) * total_admitidos
df['Tasa de Admisión (%)'] = (df['Admitidos'] / df['Aspirantes']) * 100

# 2. Representación Gráfica con Plotly del análisis de las afirmaciones
fig = go.Figure()

# Barra de Aspirantes
fig.add_trace(go.Bar(
    x=df['Programa'],
    y=df['Aspirantes'],
    name='Total Aspirantes',
    marker_color='lightslategrey',
    text=df['Aspirantes'],
    textposition='auto'
))

# Barra de Admitidos
fig.add_trace(go.Bar(
    x=df['Programa'],
    y=df['Admitidos'],
    name='Admitidos Calculados',
    marker_color='crimson',
    text=df['Admitidos'].astype(int),
    textposition='auto'
))

# Línea interactiva secundaria para mostrar la tasa real de ingreso
fig.add_trace(go.Scatter(
    x=df['Programa'],
    y=df['Tasa de Admisión (%)'] * 100, # Escalado para visibilidad en el eje común o informativo
    mode='lines+markers',
    name='Tasa de Ingreso (Info)',
    text=df['Tasa de Admisión (%)'].round(1).astype(str) + '%',
    hoverinfo='text',
    line=dict(color='gold', width=3)
))

# Ajustes de diseño interactivo
fig.update_layout(
    title='<b>Análisis Comparativo: Aspirantes vs Admitidos por Programa</b><br>Evaluación de relaciones para verificar la solución',
    xaxis_title='Programas Académicos',
    yaxis_title='Cantidad de Personas',
    barmode='group',
    template='plotly_white'
)

fig.show()
fig.write_image(r"C:\Users\marco\Downloads\analisis_comparativo_aspirantes_admitidos.png", format='png', scale=2)
````

## Solución matemática del problema

Para resolver el problema, recordemos primero los datos clave deducidos de los gráficos anteriores:

* **Aspirantes (Gráfico de barras):**
* Ciencias (C): $3000$ (3 mil)
* Medicina (M): $6000$ (6 mil)


* **Admitidos (Gráfico circular):**
* Total de admitidos = $3600$
* Ángulo de Ciencias (C) = $20^\circ \implies \text{Admitidos}_C = \frac{20}{360} \times 3600 = 200$
* Ángulo de Medicina (M) = $70^\circ \implies \text{Admitidos}_M = \frac{70}{360} \times 3600 = 700$



El problema nos pide analizar la **razón entre aspirantes y admitidos**, es decir:


$$\text{Razón} = \frac{\text{Aspirantes}}{\text{Admitidos}}$$

Calculemos las razones individuales para cada programa:

1. **Razón en Ciencias (C):**

$$\text{Razón}_C = \frac{3000}{200} = 15$$



*(Esto significa que por cada 15 aspirantes a Ciencias, ingresa 1)*.
2. **Razón en Medicina (M):**

$$\text{Razón}_M = \frac{6000}{700} = \frac{60}{7} \approx 8.57$$



*(Esto significa que por cada ~8.57 aspirantes a Medicina, ingresa 1)*.

**Comparación:**
Al comparar los resultados observamos que $15 > 8.57$, por lo tanto, **la razón de Ciencias es mayor que la de Medicina**.

* **Respuesta correcta:** **(c) La de ciencias es mayor que la de medicina**


# Solución matemática del problema

Evaluemos minuciosamente cada una de las opciones para identificar cuál de ellas es **FALSA**:



* **Evaluación de (a):** *"En el año 2000 se vendieron 675 toneladas de metal."*
* Ventas del año 2000 = $\text{Aluminio} + \text{Hierro} + \text{Cobre} = 300 + 150 + 225 = 675\text{ Toneladas}$.
* Esta afirmación es **VERDADERA**.




* **Evaluación de (b):** *"La cantidad total de metal vendido en el año 1990 fue igual a la vendida en el año 1995."*
* Total 1990 = $200 + 125 + 100 = 425\text{ Toneladas}$.
* Total 1995 = $200 + 75 + 150 = 425\text{ Toneladas}$.
* Dado que $425 = 425$, esta afirmación es **VERDADERA**.




* **Evaluación de (c):** *"La cantidad total de metal vendido en el año 1990 tuvo un incremento del 20% con respecto al año 1985."*
* Total 1985 = $100 + 125 + 75 = 300\text{ Toneladas}$.
* Incremento porcentual de 1990 respecto a 1985:

$$\text{Incremento} = \frac{\text{Total 1990} - \text{Total 1985}}{\text{Total 1985}} \times 100 = \frac{425 - 300}{300} \times 100 = \frac{125}{300} \times 100 \approx 41.67\%$$

## Solución matemática del problema

Para resolver el problema, analicemos las dos sucesiones aritméticas que rigen las coordenadas de los extremos de los segmentos de orden $n$:

### 1. Análisis de la coordenada $X$:

La sucesión para los valores de $X$ es: $-5, -4.5, -4, -3.5, -3, \dots$

* Es una progresión aritmética con término inicial $X_1 = -5$.
* La diferencia común ($d_x$) es: $-4.5 - (-5) = +0.5$.
* La fórmula para el término general $n$ es:

$$X_n = X_1 + (n - 1) \cdot d_x = -5 + (n - 1) \cdot 0.5$$



Sustituyendo para el **orden 10** ($n = 10$):


$$X_{10} = -5 + (10 - 1) \cdot 0.5 = -5 + 9 \cdot 0.5 = -5 + 4.5 = -0.5$$

### 2. Análisis de la coordenada $Y$:

La sucesión para los valores de $Y$ es: $1, 1.5, 2, 2.5, 3, \dots$

* Es una progresión aritmética con término inicial $Y_1 = 1$.
* La diferencia común ($d_y$) es: $1.5 - 1 = +0.5$.
* La fórmula para el término general $n$ es:

$$Y_n = Y_1 + (n - 1) \cdot d_y = 1 + (n - 1) \cdot 0.5$$



Sustituyendo para el **orden 10** ($n = 10$):


$$Y_{10} = 1 + (10 - 1) \cdot 0.5 = 1 + 9 \cdot 0.5 = 1 + 4.5 = 5.5$$

Las coordenadas para el segmento de orden 10 son **$-0.5$ y $5.5$**.
