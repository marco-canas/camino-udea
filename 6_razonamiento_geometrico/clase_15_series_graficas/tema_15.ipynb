```python
import qrcode
from IPython.display import display
from PIL import Image

def generar_qr(url, version=1, box_size=4, border=4, fill_color="black", back_color="white"):
    """
    Genera y muestra un código QR para una URL dada en un Jupyter Notebook.
    
    Parámetros:
    - url: La URL para la cual generar el código QR
    - version: Tamaño del QR (1-40, donde 1 es el más pequeño)
    - box_size: Número de píxeles por cada "caja" del QR
    - border: Grosor del borde blanco alrededor del QR (en cajas)
    - fill_color: Color del código QR
    - back_color: Color de fondo del código QR
    
    Retorna:
    - Muestra el código QR directamente en el notebook
    - Retorna el objeto QR generado
    """
    # Configurar el generador de QR
    qr = qrcode.QRCode(
        version=version,
        error_correction=qrcode.constants.ERROR_CORRECT_L,
        box_size=box_size,
        border=border,
    )
    
    # Añadir la URL al QR
    qr.add_data(url)
    qr.make(fit=True)
    
    # Crear la imagen del QR
    img = qr.make_image(fill_color=fill_color, back_color=back_color)
     
    return img

# Ejemplo de uso:
# generar_qr("https://www.ejemplo.com")

```

# Solución matemática del problema

Vamos a evaluar matemáticamente cada afirmación para determinar su veracidad:

* **Análisis de la Afirmación I:** *"A ingeniería ingresaron 800 aspirantes."*
* Geométricamente, el sector **I** representa un cuadrante completo del círculo de admitidos menos 22.5°, equivalente a un ángulo de $90^\circ - 22.5^\circ = 67.5$. 


* La proporción de admitidos de Ingeniería es: $\frac{67.5^\circ}{360^\circ} = 0.1875$ ($25\%$).


```python
67.5/360
```

* Número de admitidos en Ingeniería: $3600 \times 0.1875 = 675$.


```python
3600*0.1875
```

* Como $675 \neq 800$, la afirmación **I es FALSA**.




* **Análisis de la Afirmación II:** *"De cada 10 aspirantes a estudiar artes fueron admitidos 3."*
* El sector **A** también representa un cuadrante de $90^\circ$ ($25\%$).
* Número de admitidos en Artes: $3600 \times 0.25 = 900$.
* Del gráfico de barras sabemos que los aspirantes a Artes son 3 mil ($3000$).
* Relación de admisión para Artes: $\frac{\text{Admitidos}}{\text{Aspirantes}} = \frac{900}{3000} = \frac{3}{10}$.
* Esto significa exactamente que ingresan 3 de cada 10 aspirantes. Por lo tanto, la afirmación **II es VERDADERA**.


* **Análisis de la Afirmación III:** *"El porcentaje de admitidos con relación a aspirantes por programa es igual en artes que en derecho."*
* Ya calculamos la relación para Artes, la cual es del $30\%$.
* Para Derecho (**D**), el gráfico circular muestra otro cuadrante completo de $90^\circ$ ($25\%$), lo que equivale a $900$ admitidos.
* Del gráfico de barras vemos que Derecho tiene 4 mil ($4000$) aspirantes.
* Relación de admisión para Derecho: $\frac{900}{4000} = \frac{9}{40} = 0.225$ ($22.5\%$).
* Como $30\% \neq 22.5\%$, la afirmación **III es FALSA**.



Concluimos que **Solo la II es verdadera**. La respuesta correcta es la **(a)**.

## Representación del proceso de solución con Pandas y Plotly

Este script en Python utiliza **Pandas** para procesar las tablas de datos y **Plotly** para generar una visualización interactiva que compara directamente las tasas de admisión de los programas evaluados.



```python
import pandas as pd
import plotly.graph_objects as go

# 1. Crear el set de datos estructurado en un DataFrame de Pandas
datos = {
    'Programa': ['Ingeniería (I)', 'Artes (A)', 'Medicina (M)', 'Ciencias (C)', 'Derecho (D)'],
    'Aspirantes': [5000, 3000, 6000, 3000, 4000],
    'Ángulo Circular': [90-22.5, 90, 22.5, 22.5, 180 - 22.5] 
}

df = pd.DataFrame(datos)

# Calcular admitidos y tasa de admisión (porcentaje de ingreso por programa)
total_admitidos = 3600
df['Admitidos'] = (df['Ángulo Circular'] / 360) * total_admitidos
df['Tasa de Admisión (%)'] = (df['Admitidos'] / df['Aspirantes']) * 100

# 2. Representación Gráfica con Plotly del análisis de las afirmaciones
fig = go.Figure()

# Barra de Aspirantes
fig.add_trace(go.Bar(
    x=df['Programa'],
    y=df['Aspirantes'],
    name='Total Aspirantes',
    marker_color='lightslategrey',
    text=df['Aspirantes'],
    textposition='auto'
))

# Barra de Admitidos
fig.add_trace(go.Bar(
    x=df['Programa'],
    y=df['Admitidos'],
    name='Admitidos Calculados',
    marker_color='crimson',
    text=df['Admitidos'].astype(int),
    textposition='auto'
))

# Línea interactiva secundaria para mostrar la tasa real de ingreso
fig.add_trace(go.Scatter(
    x=df['Programa'],
    y=df['Tasa de Admisión (%)'] * 100, # Escalado para visibilidad en el eje común o informativo
    mode='lines+markers',
    name='Tasa de Ingreso (Info)',
    text=df['Tasa de Admisión (%)'].round(1).astype(str) + '%',
    hoverinfo='text',
    line=dict(color='gold', width=3)
))

# Ajustes de diseño interactivo
fig.update_layout(
    title='<b>Análisis Comparativo: Aspirantes vs Admitidos por Programa</b><br>Evaluación de relaciones para verificar la solución',
    xaxis_title='Programas Académicos',
    yaxis_title='Cantidad de Personas',
    barmode='group',
    template='plotly_white'
)

fig.show()
fig.write_image(r"C:\Users\marco\Downloads\analisis_comparativo_aspirantes_admitidos.png", format='png', scale=2)
```

### Implementing MLPs with Keras

Keras is a high-level Deep Learning API that allows you to easily build, train, evaluate, and execute all sorts of neural networks. Its documentation (or specification) is available at https://keras.io/. The reference implementation, also called Keras, was developed by François Chollet as part of a research project and was released as an open source project in
March 2015. It quickly gained popularity, owing to its ease of use, flexibility, and beautiful design. To perform the heavy computations required by neural networks, this reference implementation relies on a computation backend. At present, you can choose from three popular open source Deep Learning libraries: TensorFlow, Microsoft Cognitive Toolkit(CNTK), and Theano. Therefore, to avoid any confusion, we will refer to this reference implementation as multibackend Keras. Since late 2016, other implementations have been released. You can now run Keras on Apache MXNet, Apple’s Core ML, JavaScript or TypeScript (to run Keras code in a web browser), and PlaidML (which can run on all sorts of GPU devices, not just Nvidia). Moreover, TensorFlow itself now comes bundled with its own Keras implementation, tf.keras. It only supports TensorFlow as the backend, but it has the advantage of offering some very useful extra features (see Figure 10-10): for example, it supports TensorFlow’s Data API, which makes it easy to load and preprocess data efficiently. For this reason, we will use tf.keras in this book. However, in this chapter we will not use any of the TensorFlow-specific features, so the code should run fine on other Keras implementations as well (at least in Python), with only minor modifications, such as changing the imports.
