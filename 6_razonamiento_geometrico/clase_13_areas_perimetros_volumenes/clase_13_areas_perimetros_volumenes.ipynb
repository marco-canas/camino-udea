````python
import matplotlib.pyplot as plt
import numpy as np

# Configurar la figura y el eje 3D
fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(111, projection='3d')

# Crear una rejilla de 3x3x3 para posicionar los cubos
# El espacio vacío será False (0) y los cubos serán True (1)
voxels = np.zeros((3, 3, 3), dtype=bool)

# Posicionar los 7 cubos en una estructura de cruz (coordenadas x, y, z)
# Centro: (1, 1, 1)
voxels[1, 1, 1] = True  # Cubo central
voxels[2, 1, 1] = True  # Derecha
voxels[0, 1, 1] = True  # Izquierda
voxels[1, 2, 1] = True  # Atrás
voxels[1, 0, 1] = True  # Adelante
voxels[1, 1, 2] = True  # Arriba
voxels[1, 1, 0] = True  # Abajo

# Configurar colores (un gris suave con bordes negros bien definidos)
colors = np.empty(voxels.shape, dtype=object)
colors[voxels] = '#f0f0f0' 

# Dibujar los voxels (cubos)
ax.voxels(voxels, facecolors=colors, edgecolors='black', linewidth=1)

# Ajustar los límites de los ejes para centrar la figura
ax.set_xlim(0, 3)
ax.set_ylim(0, 3)
ax.set_zlim(0, 3)

# Ocultar los ejes y las líneas de fondo para emular el dibujo del libro
ax.set_axis_off()

# Ajustar el ángulo de visión de la cámara para que coincida con la perspectiva isométrica
ax.view_init(elev=25, azim=45)

# guardar el gráfico
plt.savefig(r"C:\Users\marco\Downloads\figura_3d_pregunta_1.png")
# Mostrar el gráfico
plt.show()
````


# Solución detallada

Para encontrar la razón requerida, debemos calcular de manera independiente el **volumen total** y el **área superficial** de la estructura conformada por los 7 cubos unitarios.

# 1. Cálculo del volumen ($V$)

Dado que la figura está construida uniendo $7$ cubos unitarios (cada uno con un volumen de $1 \text{ u}^3$):


$$V = 7 \times 1 = 7 \text{ unidades cúbicas}$$



# 2. Cálculo del área superficial ($A$)

La estructura consiste en un **cubo central** que tiene un cubo pegado en cada una de sus $6$ caras (arriba, abajo, izquierda, derecha, adelante y atrás).

* Cada uno de los $6$ cubos periféricos tiene $6$ caras cuadradas.
* Al estar pegados al cubo central, **$1$ cara de cada cubo periférico queda oculta** (la que hace contacto con el centro).
* Por lo tanto, cada cubo periférico expone hacia el exterior exactamente $5$ de sus caras.
* El cubo central queda completamente cubierto por los demás cubos, por lo que aporta $0$ caras a la superficie exterior.

Calculamos el total de caras expuestas:


$$A = 6 \text{ cubos externos} \times 5 \text{ caras expuestas por cubo} = 30 \text{ unidades cuadradas}$$



### 3. Razón entre volumen y superficie

La razón solicitada es $\frac{Volumen}{Superficie}$:


$$\text{Razón} = \frac{V}{A} = \frac{7}{30}$$

**Respuesta correcta:** **(b) 7/30**

# Script de Matplotlib para dibujar el diagrama

Este script utiliza geometría básica y parches (`Patches`) de `matplotlib` para dibujar el sistema de poleas y las correas (abiertas y cruzadas).


````python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np

# Configurar la figura
fig, ax = plt.subplots(figsize=(7, 6))

# Definir centros y radios de las 4 ruedas
# Rueda: (x, y, radio, etiqueta)
ruedas = {
    1: (1.5, 2.0, 0.4, '1'),
    2: (2.5, 3.5, 0.2, '2'),
    3: (3.2, 2.5, 0.25, '3'),
    4: (4.2, 4.0, 0.2, '4')
}

# 1. Dibujar las ruedas (círculos)
for num, (x, y, r, label) in ruedas.items():
    circle = plt.Circle((x, y), r, edgecolor='black', facecolor='none', linewidth=1.5)
    ax.add_patch(circle)
    ax.text(x, y, label, ha='center', va='center', fontsize=12, fontweight='bold')

# Función auxiliar para dibujar líneas tangentes externas (correas abiertas)
def dibujar_correa_abierta(r1_key, r2_key):
    x1, y1, r1, _ = ruedas[r1_key]
    x2, y2, r2, _ = ruedas[r2_key]
    
    # Ángulo de la línea que une los centros
    gamma = np.arctan2(y2 - y1, x2 - x1)
    # Ángulo de las tangentes exteriores
    dist = np.hypot(x2 - x1, y2 - y1)
    alpha = np.arccos((r1 - r2) / dist)
    
    for signo in [-1, 1]:
        theta = gamma + signo * alpha
        x_start = x1 + r1 * np.cos(theta)
        y_start = y1 + r1 * np.sin(theta)
        x_end = x2 + r2 * np.cos(theta)
        y_end = y2 + r2 * np.sin(theta)
        ax.plot([x_start, x_end], [y_start, y_end], color='black', linewidth=1.5)

# Función auxiliar para dibujar líneas tangentes internas (correas cruzadas)
def dibujar_correa_cruzada(r1_key, r2_key):
    x1, y1, r1, _ = ruedas[r1_key]
    x2, y2, r2, _ = ruedas[r2_key]
    
    gamma = np.arctan2(y2 - y1, x2 - x1)
    dist = np.hypot(x2 - x1, y2 - y1)
    alpha = np.arccos((r1 + r2) / dist)
    
    # Línea superior cruzada
    theta1 = gamma + alpha
    theta2 = gamma + alpha + np.pi
    ax.plot([x1 + r1 * np.cos(theta1), x2 + r2 * np.cos(theta2)], 
            [y1 + r1 * np.sin(theta1), y2 + r2 * np.sin(theta2)], color='black', linewidth=1.5)
            
    # Línea inferior cruzada
    theta1 = gamma - alpha
    theta2 = gamma - alpha - np.pi
    ax.plot([x1 + r1 * np.cos(theta1), x2 + r2 * np.cos(theta2)], 
            [y1 + r1 * np.sin(theta1), y2 + r2 * np.sin(theta2)], color='black', linewidth=1.5)

# 2. Dibujar las correas siguiendo el diseño físico
dibujar_correa_abierta(1, 2)
dibujar_correa_cruzada(2, 3)
dibujar_correa_abierta(3, 4)

# 3. Dibujar la flecha indicadora de giro en la Rueda 1
arrow = patches.FancyArrowPatch((0.9, 2.0), (1.1, 1.4), connectionstyle="arc3,rad=-0.3",
                                 arrowstyle="->", mutation_scale=15, color='black', linewidth=2)
ax.add_patch(arrow)

# Configuraciones de los ejes
ax.set_xlim(0.5, 5.0)
ax.set_ylim(1.0, 4.5)
ax.set_aspect('equal')
ax.axis('off')

plt.savefig(r"C:\Users\marco\Downloads\pregunta_2_poleas.jpeg")

# Mostrar gráfico
plt.show()


````


## Solución del problema

Para determinar el sentido de giro de cada rueda, debemos aplicar dos principios básicos de la transmisión por poleas:

1. **Correa abierta (paralela):** Las dos ruedas unidas giran en el **mismo sentido**.
2. **Correa cruzada:** Las dos ruedas unidas giran en **sentidos opuestos**.

Analicemos la secuencia partiendo de la rueda 1:

* **Rueda 1:** Gira en el sentido de las manecillas del reloj (**Horario**), según el enunciado.
* **Rueda 2:** Está conectada a la rueda 1 mediante una correa abierta. Por lo tanto, mantiene la misma dirección. Gira en sentido **Horario**.
* **Rueda 3:** Está conectada a la rueda 2 por una correa cruzada. Al estar cruzada, invierte la dirección de giro. Por lo tanto, gira en sentido **Antihorario** (contrario a las manecillas).
* **Rueda 4:** Está conectada a la rueda 3 mediante una correa abierta. Conserva la dirección de la rueda 3, por lo que también gira en sentido **Antihorario**.

Nos preguntan cuáles de las *otras* ruedas se mueven también en el sentido de las manecillas del reloj (Horario). La única que cumple con esto es la **rueda 2**.

**Respuesta correcta:** **(c) Sólo 2**

````python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import numpy as np

# --- CONFIGURACIÓN GENERAL DE LA FIGURA ---
fig = plt.figure(figsize=(11, 7))
fig.suptitle("Enunciado Explicativo CORREGIDO: Representación y Vistas Reales", fontsize=14, fontweight='bold', y=0.98)

# Matriz de distribución según la imagen (Planta de ejemplo)
# Fila 0 (atrás): [1, 2, 3]
# Fila 1 (medio): [1, 1, 1]
# Fila 2 (frente): [2, 2, 1]
tabla_formacion = [
    [1, 2, 3],
    [1, 1, 1],
    [2, 2, 1]
]

gs = gridspec.GridSpec(2, 4, height_ratios=[1.2, 1], figure=fig)

# ==========================================
# 1. GRÁFICO 3D (ISOMÉTRICO REAL)
# ==========================================
ax_3d = fig.add_subplot(gs[0, 0:2], projection='3d')
ax_3d.set_title("Arreglo de Cubos Fiel (3D)", fontsize=11, fontweight='bold', pad=10)

voxels = np.zeros((3, 3, 3), dtype=bool)

# Construir el cuerpo usando las alturas de la tabla (invirtiendo filas para el eje Y)
for fila_idx in range(3):
    for col_idx in range(3):
        altura = tabla_formacion[fila_idx][col_idx]
        x = col_idx
        y = 2 - fila_idx  # Mapear la fila al eje Y profundo
        voxels[x, y, 0:altura] = True

colors = np.empty(voxels.shape, dtype=object)
colors[voxels] = '#f5f5f5'
ax_3d.voxels(voxels, facecolors=colors, edgecolors='black', linewidth=0.8)

ax_3d.set_xlim(0, 3)
ax_3d.set_ylim(0, 3)
ax_3d.set_zlim(0, 3)
ax_3d.view_init(elev=22, azim=-45) # Ángulo ideal para ver la elevación escalonada trasera
ax_3d.axis('off')


# ==========================================
# 2. TABLA DE FORMACIÓN EN PERSPECTIVA (PLANTA)
# ==========================================
ax_planta = fig.add_subplot(gs[0, 2:4])
ax_planta.set_title("Tabla de Formación (Planta Real)", fontsize=11, fontweight='bold')

for i in range(4):
    ax_planta.plot([0 + i*0.3, 3 + i*0.3], [i, i], color='black', linewidth=1)
    ax_planta.plot([i, i + 3*0.3], [0, 3], color='black', linewidth=1)

# Renderizar los caracteres exactamente en su posición matricial
for fila in range(3):
    for col in range(3):
        x_pos = col + 0.5 + (fila + 0.5) * 0.3
        y_pos = fila + 0.5
        # Invertimos el orden visual para que coincida con el dibujo del papel
        ax_planta.text(x_pos, y_pos, str(tabla_formacion[2 - fila][col]), 
                       ha='center', va='center', fontsize=11, fontweight='bold')

ax_planta.set_xlim(-0.5, 4.5)
ax_planta.set_ylim(-0.5, 3.5)
ax_planta.set_aspect('equal')
ax_planta.axis('off')


# ==========================================
# FUNCIÓN PARA LAS VISTAS PLANAS 2D
# ==========================================
def dibujar_vista_exacta(ax, alturas, nombre_vista):
    for i in range(4):
        ax.plot([0, 3], [i, i], color='lightgray', linewidth=0.5)
        ax.plot([i, i], [0, 3], color='lightgray', linewidth=0.5)
    
    for col, altura in enumerate(alturas):
        for fila in range(altura):
            rect = plt.Rectangle((col, fila), 1, 1, facecolor='#fafafa', edgecolor='black', linewidth=1.2)
            ax.add_patch(rect)
            
    ax.set_xlim(-0.1, 3.1)
    ax.set_ylim(-0.1, 3.1)
    ax.set_aspect('equal')
    ax.axis('off')
    ax.set_title(nombre_vista, fontsize=10, y=-0.25, fontweight='bold')


# ==========================================
# 3. LAS 4 VISTAS COHERENTES CON LA MATRIZ
# ==========================================
# Mapeo corregido de la vista lateral derecha según la perspectiva del papel
vistas_correctas = {
    0: ([2, 2, 3], "Frontal"),
    1: ([2, 1, 3], "Lateral\nderecha"),
    2: ([3, 2, 2], "Posterior"),
    3: ([3, 1, 2], "Lateral\nizquierda")
}

for idx, (alturas, nombre) in vistas_correctas.items():
    ax_vista = fig.add_subplot(gs[1, idx])
    dibujar_vista_exacta(ax_vista, alturas, nombre)

plt.tight_layout()
plt.savefig(r"C:\Users\marco\Downloads\pregunta_3_vistas_reales.png")
plt.show()
````

# Script de Matplotlib usando `subplots()`

Para representar las opciones de respuesta en una matriz gráfica clara de $2 \times 2$, utilizaremos `plt.subplots(2, 2)`. El script dibuja las cuadrículas correspondientes a las opciones **(a), (b), (c) y (d)** de la pregunta 13.3, asignando un título a cada una y un título global a la figura entera (`suptitle`).


````python
import matplotlib.pyplot as plt
import numpy as np

# Definir una función para dibujar las vistas de cuadrículas de 3x3 celdas fijas
def dibujar_vista(ax, alturas, opcion_letra):
    # Crear la cuadrícula base de fondo (3 filas x 3 columnas)
    for i in range(4):
        ax.plot([0, 3], [i, i], color='gray', linestyle='-', linewidth=0.8) # Líneas horizontales
        ax.plot([i, i], [0, 3], color='gray', linestyle='-', linewidth=0.8) # Líneas verticales

    # Dibujar los bloques rellenos (vistas) según la altura máxima visible de cada columna
    for col, altura in enumerate(alturas):
        for fila in range(altura):
            # Dibujar el contorno exterior en negro para las celdas ocupadas
            rect = plt.Rectangle((col, fila), 1, 1, facecolor='#f7f7f7', edgecolor='black', linewidth=1.5)
            ax.add_patch(rect)
            
    # Configurar límites y diseño del subgráfico
    ax.set_xlim(-0.2, 3.2)
    ax.set_ylim(-0.2, 3.2)
    ax.set_aspect('equal')
    ax.axis('off') # Ocultar ejes coordenados numéricos
    
    # Título individual para cada gráfico secundario (Subplot)
    ax.set_title(f"Opción ({opcion_letra})", fontsize=11, pad=8, fontweight='bold')

# 1. Definir los datos de las alturas de las columnas de las opciones (a, b, c, d)
# Leyendo las columnas de izquierda a derecha según la imagen de la pregunta 13.3
opciones = {
    'a': [2, 3, 3],
    'b': [2, 3, 3],
    'c': [3, 3, 2],
    'd': [3, 2, 2]
}

# 2. Inicializar la matriz de gráficos con subplots() (2 filas, 2 columnas)
fig, axs = plt.subplots(2, 2, figsize=(7, 7))

# 3. Título general para toda la figura entera
fig.suptitle("Pregunta 13.3: Vistas Laterales del Arreglo", fontsize=14, fontweight='bold', y=0.98)

# 4. Iterar sobre la matriz plana para construir cada opción
dibujar_vista(axs[0, 0], opciones['a'], 'a')
dibujar_vista(axs[0, 1], opciones['b'], 'b')
dibujar_vista(axs[1, 0], opciones['c'], 'c')
dibujar_vista(axs[1, 1], opciones['d'], 'd')

# Ajustar los espacios de manera limpia y organizada
plt.tight_layout(rect=[0, 0.03, 1, 0.95])

# Mostrar el arreglo de gráficos
plt.show()
````

## 📋 Enunciado de la pregunta

> **13.5.** Juan dispone de un trozo rectangular de cartulina de $12\text{ cm}$ de largo, para recortar $7$ círculos iguales de $3\text{ cm}$ de diámetro. La altura del rectángulo de cartulina de menor área para que Juan pueda ejecutar esta labor, en $\text{cm}$, es:

> * **(d)** $6$

## 🧠 Solución geométrica paso a paso

Para minimizar el área de la cartulina manteniendo fijo el largo en $12\text{ cm}$, debemos acomodar los círculos usando un **empaquetamiento compacto (hexagonal u oblicuo)** en lugar de un empaquetamiento ortogonal (en cuadrícula alineada), ya que el empaquetamiento hexagonal reduce drásticamente la altura total requerida.

### 1. Distribución de los círculos

* Cada círculo tiene un diámetro $d = 3\text{ cm}$, por lo que su radio es $r = 1.5\text{ cm}$.
* Disponemos de un largo total de $12\text{ cm}$. Si alineamos los círculos horizontalmente en una fila estándar, caben exactamente:

$$\frac{12\text{ cm}}{3\text{ cm}} = 4 \text{ círculos por fila}$$



Para acomodar los $7$ círculos de la forma más baja posible, creamos dos filas intercaladas:

* **Fila inferior:** Contiene $4$ círculos colocados uno al lado del otro. Ocupan exactamente los $12\text{ cm}$ de largo ($4 \times 3 = 12$).
* **Fila superior:** Contiene los $3$ círculos restantes, colocados en los huecos o valles formados entre los círculos de la primera fila.

### 2. Distancia vertical entre los centros de las filas

Al intercalar los círculos de esta forma, los centros de tres círculos adyacentes que están en contacto mutuo forman un **triángulo equilátero**.

* El lado de este triángulo equilátero es igual a la suma de dos radios: $L = r + r = d = 3\text{ cm}$.
* La distancia vertical ($h_{centro}$) entre el centro de la fila inferior y el centro de la fila superior equivale a la altura de dicho triángulo equilátero.

Usando el teorema de Pitágoras (o la propiedad de los triángulos de $30^\circ-60^\circ-90^\circ$):


$$h_{centro} = L \cdot \frac{\sqrt{3}}{2} = 3 \cdot \frac{\sqrt{3}}{2} = \frac{3}{2}\sqrt{3}\text{ cm}$$

### 3. Altura total del rectángulo de cartulina

La altura total del rectángulo debe cubrir toda la extensión vertical de los círculos:

1. Desde el borde inferior del rectángulo hasta el centro de la primera fila hay una distancia de $1$ radio ($r$).
2. Desde el centro de la primera fila hasta el centro de la segunda fila hay una distancia de $h_{centro}$.
3. Desde el centro de la segunda fila hasta el borde superior del rectángulo hay otra distancia de $1$ radio ($r$).

$$\text{Altura Total} = r + h_{centro} + r$$

$$\text{Altura Total} = 2r + h_{centro}$$

Como sabemos que $2r = d = 3\text{ cm}$:


$$\text{Altura Total} = 3 + \frac{3}{2}\sqrt{3}\text{ cm}$$

Reorganizando los términos para que coincida con las opciones:


$$\text{Altura Total} = \frac{3}{2}\sqrt{3} + 3$$

**Respuesta correcta:** **(b) $\frac{3}{2}\sqrt{3} + 3$**

## 🐍 Script de Matplotlib para graficar la solución

Este script dibuja de manera exacta la cartulina rectangular de $12\text{ cm}$ de largo y la altura mínima calculada, mostrando la posición geométrica precisa de los $7$ círculos y el triángulo equilátero que define el problema.

```python


````python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np

# Datos del problema
d = 3.0       # Diámetro
r = d / 2.0   # Radio (1.5 cm)
largo = 12.0  # Largo fijo de la cartulina

# Cálculos geométricos del empaquetamiento compacto
h_centro = d * (np.sqrt(3) / 2) # Altura entre centros (3/2 * sqrt(3))
altura_total = d + h_centro     # Altura de la cartulina (3 + 3/2 * sqrt(3))

# Configurar la gráfica
fig, ax = plt.subplots(figsize=(10, 5))

# 1. Dibujar el rectángulo de la cartulina de menor área
cartulina = patches.Rectangle((0, 0), largo, altura_total, linewidth=2, 
                              edgecolor='darkred', facecolor='#fff5f5', label='Cartulina Mínima')
ax.add_patch(cartulina)

# 2. Posicionar los 7 círculos
# Fila inferior: 4 círculos
centros_fila1 = [(r + i * d, r) for i in range(4)]
# Fila superior: 3 círculos intercalados en los valles
centros_fila2 = [(d + i * d, r + h_centro) for i in range(3)]

todos_los_centros = centros_fila1 + centros_fila2

for centro in todos_los_centros:
    circulo = plt.Circle(centro, r, edgecolor='blue', facecolor='#e6f2ff', alpha=0.8, linewidth=1.5)
    ax.add_patch(circulo)
    ax.plot(centro[0], centro[1], 'ko', markersize=3) # Punto del centro

# 3. Dibujar el triángulo equilátero de referencia entre los centros
# Usaremos el primer círculo de la fila superior y los dos de abajo que lo sostienen
c_izq = centros_fila1[0]
c_der = centros_fila1[1]
c_sup = centros_fila2[0]

triangulo = plt.Polygon([c_izq, c_der, c_sup], closed=True, 
                        edgecolor='green', facecolor='none', linestyle='--', linewidth=2, label='Triángulo Equilátero (Lado = 3cm)')
ax.add_patch(triangulo)

# Línea que representa h_centro (la altura del triángulo)
ax.plot([c_sup[0], c_sup[0]], [c_izq[1], c_sup[1]], color='purple', linestyle=':', linewidth=2, label='Altura entre centros ($h_{centro}$)')

# Cifras y anotaciones de cotas informativas
ax.text(largo + 0.3, altura_total / 2, f'Altura Total:\n$\\frac{{3}}{{2}}\\sqrt{{3}} + 3 \\approx {altura_total:.3f}$ cm', 
        va='center', ha='left', color='darkred', fontweight='bold', fontsize=11)
ax.text(largo / 2, -0.5, 'Largo fijo = 12 cm', va='top', ha='center', color='black', fontsize=11)

# Ajustes de los ejes de coordenadas
ax.set_xlim(-0.5, largo + 4.0)
ax.set_ylim(-1.0, altura_total + 1.0)
ax.set_aspect('equal')
ax.axis('off')
ax.set_title('Distribución Óptima Compacta de los 7 Círculos', fontsize=13, fontweight='bold', pad=10)
ax.legend(loc='lower right')

# Mostrar la imagen resultante
plt.tight_layout()
plt.show()
````
